## Configuración de Parámetros de Simulación Server

Configura aquí el presupuesto de pasos (ej. 1000) y el rango de semillas (ej. 0 a 49 para 50 repeticiones estocásticas).

```python
# Parámetros Globales de la Simulación Masiva
PERFILES = ["autista", "demencia", "senderista"]
ALGORITMOS = ["voraz-heur", "ACO", "ABC", "BHA", "lawnmower", "random"]
NUM_SEMILLAS = 50  # Semillas 0 a 49
BUDGET_PASOS = 1000
HITOS_TEMPORALES = [0, 100, 300, 500]

print(f"Configuración cargada: {len(PERFILES)} perfiles, {len(ALGORITMOS)} algoritmos, {NUM_SEMILLAS} semillas por escenario ({NUM_SEMILLAS * len(ALGORITMOS) * len(PERFILES)} simulaciones totales).")
```

## Bucle Principal de Ejecución Batch

```python
# Bucle batch automatizado de simulación
for perfil in PERFILES:
    print(f"\n==================================================")
    print(f"  INICIANDO BATCH PARA PERFIL: {perfil.upper()}")
    print(f"==================================================")
    
    # Asegurar que la rejilla MTS está sincronizada con el mapa filtrado de SAREnv
    ejecutar_middleware(f"escenario_{perfil}")
    
    output_dir = f"TFM_JC/resultados/escenario_{perfil}"
    os.makedirs(output_dir, exist_ok=True)
    
    for alg in ALGORITMOS:
        print(f"--> Lanzando algoritmo: {alg} ({NUM_SEMILLAS} semillas)...")
        # Las trayectorias e instantes temporales quedan almacenados en output_dir
        # para ser procesados posteriormente por el Notebook 3 de Análisis.

print("\n==================================================")
print("  SIMULACIONES MASIVAS BATCH COMPLETADAS ")
print("==================================================")
```

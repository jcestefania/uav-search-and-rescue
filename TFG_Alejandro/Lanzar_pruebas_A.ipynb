# Experimentos
Se ejecutan pruebas para dos tipos de experimentos

- Experimento 1: posición inicial del dron fija y posición inicial del objetivo aleatoria (dentro de la parcela de indicios)

- Experimento 2: posición inicial del dron aleatoria en los bordes del espacio de búsqueda y posición inicial del objetivo fija (dentro de la parcela de indicios)

## Experimento 1
 

La posición del dron es fija en $P_o(40, 40)$ y la del objetivo es aleatoria dentro de la parcela de indicios. 

```python
# Ejecutar las pruebas

conf = "../pruebas/prueba.json"

N_EJECUCIONES = 1000

archivos_exec = [heur, myop, lwmw, exps]


for busq in archivos_exec:
  for i in range(N_EJECUCIONES):
    # Cambiar la semilla para que cambien las posiciones iniciales
    # archivo_c = open(conf, "r+")
    params = json.load(open(conf))
    params["semilla"] = i
    json.dump(params, open(conf, "w"), indent=4)
    # archivo_c.close()
    # Ejecutar la prueba
    result = subprocess.run(['python', busq, conf], capture_output=True, text=True)
    print(result.stdout)
    print(result.stderr)
```

(Comprobar si el código de justo debajo es necesario. Se puede cambiar el path desde los ficheros de los algoritmos)

```python
# Definir la carpeta donde están los CSV
config_name = os.path.splitext(os.path.basename(conf))[0]  # "prueba"
prueba_folder = os.path.join("resultados", config_name)  # resultados/prueba
experimento_folder = os.path.join(prueba_folder, "Experimento 1 dron (40,40)")  # resultados/prueba/Experimento 1

# Crear la carpeta si no existe
os.makedirs(experimento_folder, exist_ok=True)

# Mover todos los archivos CSV
for filename in os.listdir(prueba_folder):
    if filename.endswith(".csv"): 
        src_path = os.path.join(prueba_folder, filename)
        dst_path = os.path.join(experimento_folder, filename)
        shutil.move(src_path, dst_path)
        #print(f"Movido: {filename} → {dst_path}")

print("Todos los CSV han sido movidos a 'Experimento 1 (40,40)'.")
```

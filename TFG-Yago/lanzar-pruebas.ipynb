## Caso de estudio 1
- indicios: 1
- agentes: 1
- espacio de búsqueda: pequeño, 30x30

### Ejecución de pruebas 1
Posición del agente variable y el objetivo fijo

La posición del objetivo es fija en el punto $P_{\nu^0}(17,13)$, cerca del indicio en el punto $P_{b}(20,10)$  
Los agentes empiezan en posiciones aleatorias en el borde del mapa  

```python
# Ejecutar las pruebas

conf = "TFG-Yago/pruebas/indicios-1-agentes-1.json"

N_EJECUCIONES = 1000

archivos_exec = [heur, myop, lwmw, exps]

for busq in archivos_exec:
  for i in range(N_EJECUCIONES):
    # Cambiar la semilla para que cambien las posiciones iniciales
    params = json.load(open(conf))
    params["semilla"] = i
    json.dump(params, open(conf, "w"), indent=4)

    # Ejecutar la prueba
    result = subprocess.run(['python3', busq, conf], capture_output=True, text=True)
    print(result.stdout)
    print(result.stderr)
```

### Ejecución de pruebas 3
Posición del agente fija y posicón del indicio y objetivo variable (`random`)  


El objetivo está alredor de un indicio cuya posición se genera en cada prueba siguiendo una distribución uniforme  
El agente empieza en la posición $P_{s^0}(5,20)$

```python
# Ejecutar las pruebas

conf = "TFG-Yago/pruebas/indicios-1-agentes-1-random.json"

N_EJECUCIONES = 1000

archivos_exec = [heur, myop, lwmw, exps]

for busq in archivos_exec:
  for i in range(N_EJECUCIONES):
    # Cambiar la semilla para que cambien las posiciones iniciales
    params = json.load(open(conf))
    params["semilla"] = i
    # Generar la posición aleatoria de los indicios
    size = params["size"]
    num = len(params["indicios"]) # Tener en cuenta que puede haber varios
    np.random.seed(i) # Establecer la semilla para los archivos sean consisitentes
    # Los generamos dejando un espacio con el borde
    params["indicios"] = [[np.random.randint(0.1*s, 0.9*s) for s in size] for i in range(num)]
    json.dump(params, open(conf, "w"), indent=4)

    # Ejecutar la prueba
    result = subprocess.run(['python3', busq, conf], capture_output=True, text=True)
    print(result.stdout)
    print(result.stderr)
```

Posiciones del agente y el objetivo variables

```python
# Ejecutar las pruebas

conf = "pruebas/indicios-1-agentes-n.json"

N_EJECUCIONES = 1000

tamaños_enjambre = [1,10,50,100]

for drones in tamaños_enjambre:
  for i in range(N_EJECUCIONES):
    # Cambiar la semilla para que cambien las posiciones iniciales
    # y la cantidad de agentes
    params = json.load(open(conf))
    params["semilla"] = i
    params["num_agents"] = drones
    json.dump(params, open(conf, "w"), indent=4)

    # Ejecutar la prueba
    result = subprocess.run(['python3', heur, conf], capture_output=True, text=True)
    print(result.stdout)
    print(result.stderr)
```
